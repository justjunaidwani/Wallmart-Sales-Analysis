## Wallmart Sales Data


```python

# Importing Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python

data = pd.read_csv("Walmart.csv", encoding_errors='ignore')

data.head(5)

```

```python

data.info()

```

```python
# Changing Column names to all lowercase.

data.columns = data.columns.str.lower()

```

```python
data.info()
```

- unit_price should be in float/integer
- date and time coloumn can be converted to date and time column respectively
- unit_price and quantity contains null values


```python

data['unit_price'] = data['unit_price'].str.replace('$','').astype(float)

# Converted unit price to float datatype by removing the '$' sign first.
```

```python

data['date'] = pd.to_datetime(data['date'])
data['time'] = pd.to_datetime(data['time'], format='%H:%M:%S').dt.time

# Converted date and time column to their appropriate datatypes.
```

```python
(data.isnull().sum()/data.count())*100
```

Null values in the column compromises of only 0.3% which is negligble. So, it is best to drop the null values in this case.

```python

data.dropna(inplace=True)

```

```python
data.info()
```

```python
#  Let's check for the duplicate values

data.duplicated().sum()
```

There are 51 duplicate values in the dataset.

```python

data.drop_duplicates(inplace=True)

```

```python

# Creating new column from already existing columns.

data['revenue'] = data['unit_price'] * data['quantity']

```

```python

data.head(5)

```

```python

data.to_csv("Cleaned_Walmart_Dataset.csv", index=False)

```
